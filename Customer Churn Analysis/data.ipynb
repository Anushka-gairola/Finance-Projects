
# Generating data for customer table of 500 rows 

```python
import pandas as pd
import numpy as np

# Reproducible results
np.random.seed(42)

# Number of customers
n_customers = 500

# Customer IDs
customer_ids = range(1001, 1001 + n_customers)

# Gender distribution
genders = np.random.choice(
    ['Male', 'Female'],
    size=n_customers,
    p=[0.5, 0.5]
)

# Age distribution
ages = np.random.randint(18, 65, n_customers)

# UK cities
cities = np.random.choice(
    [
        'London',
        'Manchester',
        'Birmingham',
        'Leeds',
        'Liverpool',
        'Bristol'
    ],
    size=n_customers
)

# Customer segments
segments = np.random.choice(
    ['Premium', 'Standard', 'Basic'],
    size=n_customers,
    p=[0.2, 0.5, 0.3]
)

# Join dates
join_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range(
            start='2020-01-01',
            end='2024-12-31'
        ),
        size=n_customers
    )
)

# Create DataFrame
customers = pd.DataFrame({
    'customer_id': customer_ids,
    'gender': genders,
    'age': ages,
    'city': cities,
    'join_date': join_dates,
    'customer_segment': segments
})

# Save to CSV
customers.to_csv(
    'customers.csv',
    index=False
)

print(customers.head())
print(f"\nTotal Customers: {len(customers)}")
```


# SUBSCRIPTIONS DATA GENERATION


```python


import pandas as pd
import numpy as np

# Read customers file
customers = pd.read_csv("customers.csv")

np.random.seed(42)

subscriptions = []

for _, row in customers.iterrows():

    customer_id = row["customer_id"]
    segment = row["customer_segment"]
    start_date = pd.to_datetime(row["join_date"])

    # Plan based on segment
    if segment == "Premium":
        plan_type = "Gold"
        monthly_fee = 50

        churn_probability = 0.10

    elif segment == "Standard":
        plan_type = "Silver"
        monthly_fee = 30

        churn_probability = 0.20

    else:
        plan_type = "Basic"
        monthly_fee = 20

        churn_probability = 0.35

    # Determine churn status
    churned = np.random.rand() < churn_probability

    if churned:

        status = "Churned"

        churn_date = start_date + pd.Timedelta(
            days=np.random.randint(90, 900)
        )

        churn_date = churn_date.strftime("%Y-%m-%d")

    else:

        status = "Active"
        churn_date = None

    subscriptions.append([
        2000 + len(subscriptions) + 1,
        customer_id,
        plan_type,
        monthly_fee,
        start_date.strftime("%Y-%m-%d"),
        churn_date,
        status
    ])

subscriptions_df = pd.DataFrame(
    subscriptions,
    columns=[
        "subscription_id",
        "customer_id",
        "plan_type",
        "monthly_fee",
        "start_date",
        "churn_date",
        "status"
    ]
)

subscriptions_df.to_csv(
    "subscriptions.csv",
    index=False
)

print(subscriptions_df.head())

print("\nStatus Distribution")
print(
    subscriptions_df["status"]
    .value_counts()
)
```


# TRANSACTIONS DATA GENERATION

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# =====================================================
# SUBSCRIPTIONS DATASET
#
# Creates:
# - 500 subscriptions
# - Gold / Silver / Basic plans
# - Monthly fees
# - Start dates
# - Active / Churned status
#
# Output:
# subscriptions.csv
# =====================================================

n = 500

subscription_ids = range(2001, 2001 + n)

customer_ids = range(1001, 1001 + n)

plan_types = np.random.choice(
    ["Basic", "Silver", "Gold"],
    size=n,
    p=[0.4, 0.35, 0.25]
)

monthly_fee_map = {
    "Basic": 20,
    "Silver": 30,
    "Gold": 50
}

monthly_fees = [
    monthly_fee_map[plan]
    for plan in plan_types
]

start_dates = pd.to_datetime(
    np.random.choice(
        pd.date_range(
            "2020-01-01",
            "2024-12-31"
        ),
        size=n
    )
)

status = np.random.choice(
    ["Active", "Churned"],
    size=n,
    p=[0.75, 0.25]
)

subscriptions = pd.DataFrame({
    "subscription_id": subscription_ids,
    "customer_id": customer_ids,
    "plan_type": plan_types,
    "monthly_fee": monthly_fees,
    "start_date": start_dates,
    "status": status
})

subscriptions.to_csv(
    "subscriptions.csv",
    index=False
)

print(subscriptions.head())

print("\nStatus Breakdown:")
print(
    subscriptions["status"]
    .value_counts()
)
```

# CUSTOMER SUPPORT DATA GENERATION
# ============================================

```python


import pandas as pd
import numpy as np

np.random.seed(42)

subscriptions = pd.read_csv(
    "subscriptions.csv"
)

support_tickets = []

ticket_id = 4001

issue_types = [
    "Billing",
    "Login Issue",
    "Technical Problem",
    "Payment Failure",
    "Subscription Change"
]

for _, row in subscriptions.iterrows():

    customer_id = row["customer_id"]
    status = row["status"]

    # Number of tickets per customer
    num_tickets = np.random.randint(0, 4)

    for _ in range(num_tickets):

        issue_type = np.random.choice(
            issue_types
        )

        ticket_date = pd.Timestamp(
            np.random.choice(
                pd.date_range(
                    "2022-01-01",
                    "2024-12-31"
                )
            )
        )

        if status == "Churned":

            resolution_days = np.random.randint(
                5, 11
            )

        else:

            resolution_days = np.random.randint(
                1, 4
            )

        support_tickets.append([
            ticket_id,
            customer_id,
            ticket_date.strftime(
                "%Y-%m-%d"
            ),
            issue_type,
            resolution_days
        ])

        ticket_id += 1

support_df = pd.DataFrame(
    support_tickets,
    columns=[
        "ticket_id",
        "customer_id",
        "ticket_date",
        "issue_type",
        "resolution_days"
    ]
)

support_df.to_csv(
    "customer_support.csv",
    index=False
)

print(support_df.head())

print(
    f"\nTotal Tickets: "
    f"{len(support_df)}"
)
```

```python
# ============================================
# COMPLAINTS DATA GENERATION
# ============================================

import pandas as pd
import numpy as np

np.random.seed(42)

subscriptions = pd.read_csv(
    "subscriptions.csv"
)

complaints = []

complaint_id = 5001

complaint_categories = [
    "Billing",
    "Service",
    "Technical",
    "Pricing",
    "Customer Support"
]

for _, row in subscriptions.iterrows():

    customer_id = row["customer_id"]
    status = row["status"]

    # Churned customers complain more often
    if status == "Churned":
        complaint_probability = 0.60
    else:
        complaint_probability = 0.15

    has_complaint = (
        np.random.rand()
        < complaint_probability
    )

    if has_complaint:

        # Some customers raise multiple complaints
        num_complaints = np.random.randint(
            1, 4
        )

        for _ in range(num_complaints):

            complaint_date = pd.Timestamp(
                np.random.choice(
                    pd.date_range(
                        "2022-01-01",
                        "2024-12-31"
                    )
                )
            )

            category = np.random.choice(
                complaint_categories
            )

            complaints.append([
                complaint_id,
                customer_id,
                complaint_date.strftime(
                    "%Y-%m-%d"
                ),
                category
            ])

            complaint_id += 1

complaints_df = pd.DataFrame(
    complaints,
    columns=[
        "complaint_id",
        "customer_id",
        "complaint_date",
        "complaint_category"
    ]
)

complaints_df.to_csv(
    "complaints.csv",
    index=False
)

print(complaints_df.head())

print(
    f"\nTotal Complaints: "
    f"{len(complaints_df)}"
)
```
